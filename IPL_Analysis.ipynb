```python
runs = delivery.groupby('batsman')
```

```python
top5_runs = runs['batsman_runs'].sum().sort_values(ascending=False).head(5)

top5_runs.plot(
    kind='pie',
    labels=top5_runs.index,
    autopct=lambda p: f'{int(round(p * top5_runs.sum() / 100.0))}',  # show actual runs
    startangle=90
)
```

```python
def get_batsman_runs(batsman_name):
    vk = delivery[delivery['batsman'] == batsman_name]    
    return vk.groupby('bowling_team')['batsman_runs'].sum().sort_values(ascending = False).head(5)
   
```

```python
get_batsman_runs('V Kohli')
```

```python
death_over = delivery[delivery['over'] > 15]
```

```python
all_batsman = death_over.groupby('batsman')["batsman_runs"].count().sort_values(ascending = False)
x = all_batsman > 200
batsman_list = all_batsman[x].index.tolist()
```

```python
batsman_list
```

```python
# runs scored by all these 43 batsman 
# balls played by all these 43 batsman
final = delivery[delivery['batsman'].isin(batsman_list)]
```

```python
runs = final.groupby('batsman')['batsman_runs'].sum()
```

```python
balls = final.groupby('batsman')['batsman_runs'].count()

sr = (runs/balls)*100
```

```python
sr.sort_values(ascending = False )
```
